```python
import pybryt

class Stack:

    def __init__(self):
        self.items = []

    def push(self, item):
        """將元素加入堆疊頂端"""
        self.items.append(item)

    def pop(self):
        """移除並回傳堆疊頂端的元素，若堆疊為空則回傳 None"""
        if not self.is_empty():
            return self.items.pop()
        return None

    def peek(self):
        """回傳堆疊頂端的元素但不移除，若堆疊為空則回傳 None"""
        if not self.is_empty():
            return self.items[-1]
        return None

    def is_empty(self):
        """檢查堆疊是否為空，回傳布林值"""
        return len(self.items) == 0

def is_balanced(expression):
    """利用 Stack 判斷括號是否成對匹配，回傳 True 或 False"""
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = Stack()
    for char in expression:
        if char in pairs.values():
            stack.push(char)
        elif char in pairs.keys():
            if stack.is_empty():
                return False
            top_char = stack.pop()
            if pairs[char] != top_char:
                return False
    return stack.is_empty()
import sys
from collections import deque
```

```python
std_id3n = 0
# ID 測資：std_id3n 由後端提供；直接開啟 Notebook 時可用 0 示範。
import pybryt
from copy import deepcopy as _id_copy
user_id = int(std_id3n)
pybryt.set_initial_conditions({"cq_user_id": user_id})
_id_observations = {}

def observe(checkpoint, value):
    signature = ('bonus-data-structures-3', checkpoint, user_id, _id_copy(value))
    _id_observations[checkpoint] = signature
    return signature

s = Stack()
s.push(1+user_id)
s.push(2+user_id)
observe('peek',s.peek())
observe('pop1',s.pop())
observe('pop2',s.pop())
observe('empty',s.is_empty())
for i,expression in enumerate(['([{}])','([)]','(()',''],1):
    expression = expression*(1+user_id%3)
    observe(f'balanced{i}',is_balanced(expression))
```
