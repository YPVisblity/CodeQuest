```python
import pybryt

class BSTNode:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def insert(root, data):
    if root is None:
        return BSTNode(data)
    if data < root.data:
        root.left = insert(root.left, data)
    elif data > root.data:
        root.right = insert(root.right, data)
    return root

def inorder_values(node, result=None):
    if result is None:
        result = []
    if node is not None:
        inorder_values(node.left, result)
        result.append(node.data)
        inorder_values(node.right, result)
    return result

def find_min(node):
    """回傳以 node 為根的子樹中，資料值最小的節點"""
    current = node
    while current.left is not None:
        current = current.left
    return current

def delete_node(root, data):
    """刪除 BST 中資料值等於 data 的節點，回傳刪除後的樹根"""
    if root is None:
        return root
    if data < root.data:
        root.left = delete_node(root.left, data)
    elif data > root.data:
        root.right = delete_node(root.right, data)
    else:
        if root.left is None:
            return root.right
        elif root.right is None:
            return root.left
        successor = find_min(root.right)
        root.data = successor.data
        root.right = delete_node(root.right, successor.data)
    return root
```

```python
std_id3n = 0
# ID 測資：std_id3n 由後端提供；直接開啟 Notebook 時可用 0 示範。
import pybryt
from copy import deepcopy as _id_copy
user_id = int(std_id3n)
pybryt.set_initial_conditions({"cq_user_id": user_id})
_id_observations = {}

def observe(checkpoint, value):
    signature = ('bonus-data-structures-10', checkpoint, user_id, _id_copy(value))
    _id_observations[checkpoint] = signature
    return signature

root = None
for x in [8,3,10,1,6,14,4,7]:
    root = insert(root,x+user_id)
root = delete_node(root,3+user_id)
observe('two_children',inorder_values(root))
root = delete_node(root,14+user_id)
observe('leaf',inorder_values(root))
```
