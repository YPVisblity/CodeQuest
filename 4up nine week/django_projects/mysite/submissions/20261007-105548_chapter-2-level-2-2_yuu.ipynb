```python
# 範例：處理單一例外（已完成，供參考）
def safe_open(filename):
    try:
        with open(filename) as f:
            return f.read()
    except FileNotFoundError:
        return "找不到這個檔案"

print(safe_open("不存在的檔案.txt"))
```

```python
def read_score(filename):
    try:
        with open(filename) as f:
            content = f.read().strip()
        return int(content)
    except FileNotFoundError:
        return "Error:找不到檔案"
    except ValueError:
        return "Error:格式錯誤"
```
