```python
# 正確解答
def plot_compare_investment(months, stock_values, fund_values):
    fig, ax = plt.subplots()
    ax.plot(months, stock_values, label="股票")
    ax.plot(months, fund_values, label="基金")
    ax.legend()
    return ax
```
