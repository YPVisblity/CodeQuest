```python
# 矩陣相加
# 請定義一個matrix_add(A,B)函數，計算函數相加。使用for迴圈
def matrix_add(A, B):

    result = []

    for i in range(len(A)):

        row = []

        for j in range(len(A[0])):
            row.append(A[i][j] + B[i][j])

        result.append(row)

    return result
```

```python
A = [[1,2],[3,4]]
B = [[5,6],[7,8]]
matrix_add(A,B)
A = [[2,1],[0,3]]
B = [[1,4],[5,2]]
matrix_add(A,B)
```
