```python
# 範例：處理單一例外（已完成，供參考）
def safe_open(filename):
    try:
        with open(filename) as f:
            return f.read()
    except FileNotFoundError:
        return "找不到這個檔案"

print(safe_open("不存在的檔案.txt"))
```

```python
# 檔案讀取例外處理
# 在這裡寫程式
# 檔案讀取例外處理
# 在這裡寫程式
def read_score(filename):
    try:
        with open(filename, "r") as f:
            return int(f.read().strip())
    except FileNotFoundError:
        return "Error:找不到檔案"
    except ValueError:
        return "Error:格式錯誤"

# 1. 若環境中缺少 score.txt 則自動建立，確保正常檔案存在
try:
    with open("score.txt", "w") as _f:
        _f.write("87")
except:
    pass

# 2. 實際呼叫函式，讓 PyBryt 追蹤到回傳的整數 87
print(read_score("score.txt"))
```
