```python
import pybryt
```

```python
# 正確解答
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
```

```python
grade_results = [grade(s) for s in [95, 90, 86, 80, 72, 70, 64, 60, 40]]
print(grade_results)

pybryt.Value(
    grade_results,
    name="成績等級",
    success_message="所有分數的等級都判斷正確",
    failure_message="95、90、86、80、72、70、64、60、40 應依序得到 A、A、B、B、C、C、D、D、F，請檢查邊界分數（90、80、70、60）",
)
```
