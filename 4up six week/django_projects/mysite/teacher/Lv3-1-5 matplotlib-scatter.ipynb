```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pybryt
```

```python
# 正確解答
def create_scatter_plot(height, weight):
    fig, ax = plt.subplots()

    ax.scatter(height, weight)
    ax.set_title("身高體重關係")
    ax.set_xlabel("身高")
    ax.set_ylabel("體重")

    return ax
```

```python
# 第一組資料
result_ax = create_scatter_plot(
    [150, 160, 165, 170, 180],
    [45, 55, 60, 68, 75]
)

count1 = sum(
    len(collection.get_offsets())
    for collection in result_ax.collections
)

points1 = []
for collection in result_ax.collections:
    points1.extend(
        tuple(point)
        for point in collection.get_offsets().tolist()
    )

title_result = result_ax.get_title()
xlabel_result = result_ax.get_xlabel()
ylabel_result = result_ax.get_ylabel()

# 第二組資料
result_ax2 = create_scatter_plot(
    [155, 175],
    [48, 72]
)

count2 = sum(
    len(collection.get_offsets())
    for collection in result_ax2.collections
)

points2 = []
for collection in result_ax2.collections:
    points2.extend(
        tuple(point)
        for point in collection.get_offsets().tolist()
    )

pybryt.Value(
    count1,
    name="第一組散點數量",
    success_message="正確畫出五個點",
    failure_message="第一組資料應畫出五個點"
)

pybryt.Value(
    points1,
    name="第一組散點座標",
    success_message="散點座標正確",
    failure_message="請使用傳入的 height 和 weight"
)

pybryt.Value(
    count2,
    name="第二組散點數量",
    success_message="更換資料後散點數量仍正確",
    failure_message="請勿將散點數量寫死"
)

pybryt.Value(
    points2,
    name="第二組散點座標",
    success_message="第二組散點座標正確",
    failure_message="請勿將身高和體重資料寫死"
)

pybryt.Value(title_result, name="圖表標題")
pybryt.Value(xlabel_result, name="X軸名稱")
pybryt.Value(ylabel_result, name="Y軸名稱")
```
