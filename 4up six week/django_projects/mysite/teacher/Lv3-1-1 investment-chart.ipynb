```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pybryt
```

```python
# 正確解答
def plot_investment(months,values):
    fig,ax = plt.subplots()
    ax.plot(months,values)
    ax.set_xlim(min(months), max(months))
    ax.set_ylim(min(values), max(values))
    return ax
```

```python
result_ax = plot_investment([1, 2, 3, 4, 5, 6],[100, 120, 90, 150, 200, 180])
xlim_result = result_ax.get_xlim()
ylim_result = result_ax.get_ylim()

result_ax2 = plot_investment([1, 2, 3],[50, 80, 60])
xlim_result2 = result_ax2.get_xlim()
ylim_result2 = result_ax2.get_ylim()

pybryt.Value(xlim_result, name="X軸範圍正確",
             success_message="X軸範圍正確對應到資料的月份範圍",
             failure_message="X軸範圍應為 (1, 6)，不能寫死成 (0, 12)")

pybryt.Value(ylim_result, name="Y軸範圍正確",
             success_message="Y軸範圍正確對應到資料的數值範圍",
             failure_message="Y軸範圍應為 (90, 200)")

pybryt.Value(xlim_result2, name="第二組資料X軸範圍未寫死",
             success_message="換一組資料後X軸範圍仍正確，證明沒有寫死",
             failure_message="X軸範圍應為 (1, 3)，若沿用第一組的 (1, 6) 代表座標範圍被寫死了")

pybryt.Value(ylim_result2, name="第二組資料Y軸範圍未寫死",
             success_message="換一組資料後Y軸範圍仍正確",
             failure_message="Y軸範圍應為 (50, 80)")
```
