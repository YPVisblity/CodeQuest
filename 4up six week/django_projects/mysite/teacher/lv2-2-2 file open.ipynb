```python
import pybryt
```

```python
def read_score(filename):
    try:
        with open(filename) as f:
            content = f.read().strip()
        return int(content)
    except FileNotFoundError:
        return "Error:找不到檔案"
    except ValueError:
        return "Error:格式錯誤"
```

```python
import os

result1 = read_score("不存在的檔案_xyz.txt")

bad_path = "temp_bad_score.txt"
with open(bad_path, "w") as f:
    f.write("abc")
result2 = read_score(bad_path)
os.remove(bad_path)
del bad_path

good_path = "temp_good_score.txt"
with open(good_path, "w") as f:
    f.write("6464")
result3 = read_score(good_path)
os.remove(good_path)
del good_path

good_path2 = "temp_good_score2.txt"
with open(good_path2, "w") as f:
    f.write("87")
result4 = read_score(good_path2)
os.remove(good_path2)
del good_path2

pybryt.Value(result1, name="檔案不存在處理",
             success_message="正確捕捉 FileNotFoundError",
             failure_message="檔案不存在時應回傳 'Error:找不到檔案'")

pybryt.Value(result2, name="格式錯誤處理",
             success_message="正確捕捉 ValueError",
             failure_message="內容非數字時應回傳 'Error:格式錯誤'")

pybryt.Value(result3, name="正常讀取結果",
             success_message="正確讀取並轉換分數",
             failure_message="正常檔案應回傳整數")

pybryt.Value(result4, name="讀取其他分數",
             success_message="不同檔案也能正確讀取",
             failure_message="請讀取檔案內容轉成整數，不要把分數寫死")
```
