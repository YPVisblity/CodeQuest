```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        # 情況 1：如果串列是空的，新節點直接成為 head
        if not self.head:
            self.head = new_node
            return
        # 情況 2：如果串列不是空的，走到最後一個節點並接上新節點
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def insert_at(self, index, data):
        new_node = Node(data)
        # 情況1:插入在頭部(index為0)
        if index == 0:
            new_node.next = self.head
            self.head = new_node
            return
        # 情況2:插入在中間或尾部
        current = self.head
        count = 0
        # 找到要插入位置的前一個節點(index-1)
        while current and count < index-1:
            current = current.next
            count += 1  
        # 若 current 為 None，代表 index 超出目前串列長度
        if current is None:
            raise IndexError("索引超出範圍 (Index out of bounds)")   
        # 重新串接指標
        new_node.next = current.next
        current.next = new_node

    def delete_by_value(self, data):
        # 如果串列是空的，直接結束
        if not self.head:
            return
        # 情況 1：要刪除的節點剛好是 head
        if self.head.data == data:
            self.head = self.head.next
            return

        # 情況 2：要刪除的節點在中間或尾部
        current = self.head
        # 檢查下一個節點的值是否為目標值
        while current.next:
            if current.next.data == data:
                # 跳過該節點，將指標指向下下個節點
                current.next = current.next.next
                return
            current = current.next

    def to_list(self):
        result = []
        current = self.head
        # 走訪整個串列，將資料逐一加入 Python list 中
        while current:
            result.append(current.data)
            current = current.next
            
        return result
class SinglyLinkedListAdvanced(SinglyLinkedList):
    def reverse(self):
        """將串列反轉（原地反轉節點的 next 指標，不是回傳反轉後的 list）"""
        prev = None            
        current = self.head    
        
        while current is not None:
            next_node = current.next 
            current.next = prev 
            prev = current 
            current = next_node 
            
        self.head = prev
```
