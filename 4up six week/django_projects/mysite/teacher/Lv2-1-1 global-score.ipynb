```python
import pybryt
```

```python
# 正確解答
score = 0

def update_score(points):
    global score
    score += points
    return score
```

```python
score = 0

result1 = update_score(10)
result2 = update_score(5)
result3 = update_score(-8)
result4 = update_score(0)
final_score1 = score

score_history1 = (
    result1,
    result2,
    result3,
    result4,
    final_score1,
)

score = 100

result5 = update_score(-25)
result6 = update_score(10)
final_score2 = score

score_history2 = (
    result5,
    result6,
    final_score2,
)

pybryt.Value(
    score_history1,
    name="連續更新全域分數",
    success_message="成功使用函式持續更新全域分數",
    failure_message="依序加10、加5、減8、加0後，結果應為(10, 15, 7, 7, 7)",
)

pybryt.Value(
    score_history2,
    name="不同初始分數",
    success_message="更換初始分數後仍能正確更新，沒有將分數寫死",
    failure_message="score從100開始，減25再加10後應得到(75, 85, 85)",
)
```
