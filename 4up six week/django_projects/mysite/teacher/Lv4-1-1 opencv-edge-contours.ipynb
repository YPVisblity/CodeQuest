```python
import cv2
import numpy as np
import pybryt
```

```python
# 正確解答
def draw_image_contours(image):
    # 灰階化
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # 使用Otsu方法進行二值化
    _, binary_image = cv2.threshold(gray_image,127,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU,)

    # 找出輪廓
    contours, hierarchy = cv2.findContours(
        binary_image.copy(),
        cv2.RETR_LIST,
        cv2.CHAIN_APPROX_SIMPLE,
    )

    # 在原圖副本上繪製紅色輪廓
    contour_image = image.copy()
    cv2.drawContours(contour_image,contours,-1,(0, 0, 255),2,)

    return gray_image, binary_image, contour_image
```

```python
# 測試圖片
test_image1 = np.zeros((120, 160, 3), dtype=np.uint8)
cv2.rectangle(test_image1, (20, 20), (70, 80), (255, 255, 255), -1)
cv2.circle(test_image1, (120, 60), 25, (255, 255, 255), -1)

gray1, binary1, contour1 = draw_image_contours(test_image1)

gray_signature1 = (
    gray1.shape,
    gray1.ndim,
    int(gray1.sum()),
)
binary_signature1 = (
    binary1.shape,
    tuple(int(value) for value in np.unique(binary1)),
    int(np.count_nonzero(binary1)),
)
red_mask1 = (
    (contour1[:, :, 0] == 0)
    & (contour1[:, :, 1] == 0)
    & (contour1[:, :, 2] == 255)
)
contour_signature1 = (contour1.shape, int(np.count_nonzero(red_mask1)))

pybryt.Value(
    gray_signature1,
    name="灰階轉換",
    success_message="成功將BGR圖片轉換成單通道灰階圖片",
    failure_message="請使用cv2.cvtColor與cv2.COLOR_BGR2GRAY轉換圖片",
)

pybryt.Value(
    binary_signature1,
    name="Otsu二值化",
    success_message="成功使用Otsu方法將灰階圖片二值化",
    failure_message="請使用cv2.THRESH_BINARY加上cv2.THRESH_OTSU進行二值化",
)

pybryt.Value(
    contour_signature1,
    name="紅色輪廓繪製",
    success_message="成功在原圖副本上以紅色、厚度2繪製全部輪廓",
    failure_message="請使用RETR_LIST找輪廓，並以(0, 0, 255)、厚度2繪製全部輪廓",
)
```
