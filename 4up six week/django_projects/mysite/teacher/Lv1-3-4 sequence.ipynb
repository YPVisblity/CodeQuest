```python
import pybryt
```

```python
# 正確解答
def generate_arithmetic_sequence(a, d, n):
    sequence = []
    for i in range(n):
        term = a + i * d
        sequence.append(term)
    return sequence
```

```python
val = generate_arithmetic_sequence(2, 3, 5)
val2 = generate_arithmetic_sequence(10, -4, 4)
val

pybryt.Value(
    val,
    name="等差數列",
    success_message="generate_arithmetic_sequence(2, 3, 5) 結果正確",
    failure_message="generate_arithmetic_sequence(2, 3, 5) 應回傳 [2, 5, 8, 11, 14]",
)

pybryt.Value(
    val2,
    name="不同首項與公差",
    success_message="換成其他首項與公差也正確，沒有把答案寫死",
    failure_message="generate_arithmetic_sequence(10, -4, 4) 應回傳 [10, 6, 2, -2]",
)
```
