```python
import pybryt
```

```python
# 正確解答
def create_voice_announcement(text, output_path, tts_factory):
    if not isinstance(text, str) or not text.strip():
        return "Error:文字不可為空"

    if (
        not isinstance(output_path, str)
        or not output_path.lower().endswith(".mp3")
    ):
        return "Error:檔案格式錯誤"

    tts = tts_factory(
        text=text.strip(),
        lang="zh-TW",
    )
    tts.save(output_path)

    return "語音建立完成"
```

```python
# 使用假的TTS物件評測，避免測試時連線或真的產生語音檔
tts_record = {}

class FakeTTS:
    def __init__(self, text, lang):
        tts_record["text"] = text
        tts_record["lang"] = lang

    def save(self, path):
        tts_record["saved_path"] = path

result1 = create_voice_announcement(
    "  歡迎來到第四章  ",
    "welcome.mp3",
    FakeTTS,
)
record1 = dict(tts_record)

result2 = create_voice_announcement(
    "   ",
    "empty.mp3",
    FakeTTS,
)

result3 = create_voice_announcement(
    "語音測試",
    "voice.wav",
    FakeTTS,
)

pybryt.Value(
    result1,
    name="語音建立結果",
    success_message="文字轉語音流程執行完成",
    failure_message="成功建立語音後應回傳'語音建立完成'",
)

pybryt.Value(
    record1,
    name="TTS參數與儲存路徑",
    success_message="正確傳入繁體中文文字、zh-TW語言及MP3路徑",
    failure_message="請移除文字前後空白、設定lang='zh-TW'，並呼叫save('welcome.mp3')",
)

pybryt.Value(
    result2,
    name="空白文字檢查",
    success_message="正確阻止空白文字轉換成語音",
    failure_message="文字為空或只有空白時應回傳'Error:文字不可為空'",
)

pybryt.Value(
    result3,
    name="MP3格式檢查",
    success_message="正確阻止非MP3格式的輸出",
    failure_message="輸出路徑不是.mp3時應回傳'Error:檔案格式錯誤'",
)
```
