```python
import pybryt
import numpy as np
```

```python
def safe_divide(a, b):
    try:
        return a/b
    except:
        return "ERROR"

result1 = safe_divide(10, 0)
result2 = safe_divide(10, "2")
result3 = safe_divide(10, 2)
```

```python
pybryt.Value(result1, name="除以零錯誤處理",success_message="except方法正確",failure_message="當 b=0 時應回傳 'Error:除以零'")
pybryt.Value(result2, name="型別錯誤處理",success_message="except方法正確",failure_message="當 b 不是數字時應回傳 'Error:型別錯誤'")
pybryt.Value(result3, name="正常除法結果",success_message="a除以b正確",failure_message="safe_divide(10, 2) 應回傳 5.0")
```
