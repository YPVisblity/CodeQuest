```python
import pybryt
```

```python
def matrix_max(matrix):
    max_value = matrix[0][0]
    for row in matrix:
        for number in row:
            if number > max_value:
                max_value = number
    return max_value

pybryt.ReturnValue(matrix_max([[1, 8], [3, 4]]), success_message="matrix max positive passed", failure_message="matrix max [[1,8],[3,4]] should return 8")
pybryt.ReturnValue(matrix_max([[-5, -2], [-9, -1]]), success_message="matrix max negative passed", failure_message="matrix max [[-5,-2],[-9,-1]] should return -1")
pybryt.ReturnValue(matrix_max([[7], [11], [6]]), success_message="matrix max one column passed", failure_message="matrix max [[7],[11],[6]] should return 11")
```
