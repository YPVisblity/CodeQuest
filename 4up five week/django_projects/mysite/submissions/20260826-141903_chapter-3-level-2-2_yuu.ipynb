```python
def verify_download(file_info):
    if not file_info["filename"].endswith(".mp4"):
        return "Error:格式錯誤"
    if file_info["size"] <= 0:
        return "Error:檔案損毀"
    if not file_info["checksum_valid"]:
        return "Error:校驗失敗"
    return "OK:下載成功"

# 正確解答
def verify_video_library(file_list):
    total = len(file_list)
    success = 0
    failed = []
    for f in file_list:
        result = verify_download(f)
        if result == "OK:下載成功":
            success += 1
        else:
            failed.append(f["filename"])
    return {"total": total, "success": success, "failed": failed}
```
