```python
# 正確解答
def verify_download(file_info):
    if not file_info["filename"].endswith(".mp4"):
        return "Error:格式錯誤"
    if file_info["size"] <= 0:
        return "Error:檔案損毀"
    if not file_info["checksum_valid"]:
        return "Error:校驗失敗"
    return "OK:下載成功"
result = verify_download({"filename": "lesson01.mp4", "size": 0, "checksum_valid": True})
print("AI的程式碼判定：", result)
print("但這個檔案明明格式錯誤、大小為0、校驗也失敗了")
```
