```python
# 範例：Node 類別定義（已完成，供參考）
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# 範例：走訪並印出串列內容
def print_list(head):
    current = head
    result = []
    while current is not None:
        result.append(str(current.data))
        current = current.next
    print(" -> ".join(result) if result else "(空串列)")

# 快速手動建立一個串列 1 -> 2 -> 3 來測試 print_list
n1 = Node(1)
n2 = Node(2)
n3 = Node(3)
n1.next = n2
n2.next = n3
print_list(n1)   # 預期輸出：1 -> 2 -> 3
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        # 情況 1：如果串列是空的，新節點直接成為 head
        if not self.head:
            self.head = new_node
            return
        # 情況 2：如果串列不是空的，走到最後一個節點並接上新節點
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def insert_at(self, index, data):
        new_node = Node(data)
        # 情況1:插入在頭部(index為0)
        if index == 0:
            new_node.next = self.head
            self.head = new_node
            return
        # 情況2:插入在中間或尾部
        current = self.head
        count = 0
        # 找到要插入位置的前一個節點(index-1)
        while current and count < index-1:
            current = current.next
            count += 1  
        # 若 current 為 None，代表 index 超出目前串列長度
        if current is None:
            raise IndexError("索引超出範圍 (Index out of bounds)")   
        # 重新串接指標
        new_node.next = current.next
        current.next = new_node

    def delete_by_value(self, data):
        # 如果串列是空的，直接結束
        if not self.head:
            return
        # 情況 1：要刪除的節點剛好是 head
        if self.head.data == data:
            self.head = self.head.next
            return

        # 情況 2：要刪除的節點在中間或尾部
        current = self.head
        # 檢查下一個節點的值是否為目標值
        while current.next:
            if current.next.data == data:
                # 跳過該節點，將指標指向下下個節點
                current.next = current.next.next
                return
            current = current.next

    def to_list(self):
        result = []
        current = self.head
        # 走訪整個串列，將資料逐一加入 Python list 中
        while current:
            result.append(current.data)
            current = current.next
            
        return result
```

```python
sll = SinglyLinkedList()

sll.append(10)
sll.append(20)
sll.append(30)
print(sll.to_list())

sll.insert_at(1, 15)
print(sll.to_list())

sll.delete_by_value(15)
print(sll.to_list())
```
