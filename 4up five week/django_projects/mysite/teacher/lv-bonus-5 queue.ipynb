```python
import pybryt
from collections import deque
```

```python
class Queue:
    def __init__(self):
        self.items = deque()

    def enqueue(self, item):
        """加入佇列尾端"""
        self.items.append(item)

    def dequeue(self):
        """移除並回傳佇列前端元素，若為空則回傳 None"""
        if not self.is_empty():
            return self.items.popleft()
        return None

    def peek(self):
        """查看佇列前端元素，若為空則回傳 None"""
        if not self.is_empty():
            return self.items[0]
        return None

    def is_empty(self):
        """判斷佇列是否為空"""
        return len(self.items) == 0

def simulate_print_jobs(jobs):
    """輸入工作名稱 list，依 FIFO 順序處理，回傳處理順序的 list"""
    q = Queue()

    for job in jobs:
        q.enqueue(job)
    result = []

    while not q.is_empty():
        result.append(q.dequeue())
    return result

q = Queue()
q.enqueue(1)
q.enqueue(2)
```

```python
peek_result = q.peek()
dequeue_result1 = q.dequeue()
dequeue_result2 = q.dequeue()
empty_result = q.is_empty()

jobs=["報告.pdf", "海報.pdf", "課表.pdf"]

simulate_print_jobs_result= simulate_print_jobs(jobs)

pybryt.Value(peek_result, name="peek回傳第一個元素",
             failure_message="peek() 應回傳 1")
pybryt.Value(dequeue_result1, name="dequeue第一個元素",
             failure_message="第一次 dequeue() 應回傳 1")
pybryt.Value(dequeue_result2, name="dequeue第二個元素",
             failure_message="第二次 dequeue() 應回傳 2")
pybryt.Value(empty_result, name="is_empty回傳True",
             failure_message="佇列清空後 is_empty() 應回傳 True")
pybryt.Value(
    simulate_print_jobs_result,
    name="simulate_print_jobs結果",
    failure_message="simulate_print_jobs 應依 FIFO 順序回傳正確結果"
)
```
