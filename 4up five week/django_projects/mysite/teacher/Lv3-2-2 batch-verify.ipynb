```python
import pybryt
```

```python
def verify_download(file_info):
    if not file_info["filename"].endswith(".mp4"):
        return "Error:格式錯誤"
    if file_info["size"] <= 0:
        return "Error:檔案損毀"
    if not file_info["checksum_valid"]:
        return "Error:校驗失敗"
    return "OK:下載成功"

# 正確解答
def verify_video_library(file_list):
    total = len(file_list)
    success = 0
    failed = []
    for f in file_list:
        result = verify_download(f)
        if result == "OK:下載成功":
            success += 1
        else:
            failed.append(f["filename"])
    return {"total": total, "success": success, "failed": failed}
```

```python
result1 = verify_video_library([
    {"filename": "lesson01.mp4", "size": 15420, "checksum_valid": True},
    {"filename": "lesson02.mp4", "size": 8300, "checksum_valid": True},
    {"filename": "lesson03.txt", "size": 5000, "checksum_valid": True},
    {"filename": "lesson04.mp4", "size": 0, "checksum_valid": True},
    {"filename": "lesson05.mp4", "size": 12000, "checksum_valid": False},
])

result2 = verify_video_library([
    {"filename": "a.mp4", "size": 100, "checksum_valid": True},
    {"filename": "b.mp4", "size": 200, "checksum_valid": True},
])

result3 = verify_video_library([])

pybryt.Value(result1, name="混合成功失敗的統計報告",
             success_message="正確統計出2支成功、3支失敗",
             failure_message="應為 {'total': 5, 'success': 2, 'failed': ['lesson03.txt', 'lesson04.mp4', 'lesson05.mp4']}，注意不能只用 if result: 判斷（錯誤訊息字串也是True）")

pybryt.Value(result2, name="全部成功的統計報告",
             success_message="正確統計全部成功的情況",
             failure_message="應為 {'total': 2, 'success': 2, 'failed': []}")

pybryt.Value(result3, name="空列表邊界情況",
             success_message="正確處理空影片庫的情況",
             failure_message="輸入空list時應回傳 {'total': 0, 'success': 0, 'failed': []}")
```
