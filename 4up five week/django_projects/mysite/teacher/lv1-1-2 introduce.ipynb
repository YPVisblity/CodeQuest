```python
import pybryt
```

```python
def introduce(name, age, weight):
    return "我叫" + name + "，今年" + str(age) + "歲，體重"+ str(weight) +"公斤"
```

```python
result1 = introduce("小華", 10, 32.5)
result2 = introduce("阿明", 25, 68.0)
result3 = introduce("小美", 8, 24.3)

pybryt.Value(result1, name="小華的自我介紹",
             success_message="型別轉換與字串組合正確",
             failure_message="應回傳 我叫小華，今年10歲，體重32.5公斤")

pybryt.Value(result2, name="阿明的自我介紹",
             success_message="不同數值測試正確",
             failure_message="應回傳 我叫阿明，今年25歲，體重68.0公斤")

pybryt.Value(result3, name="小美的自我介紹",
             success_message="邊界數值測試正確",
             failure_message="應回傳 我叫小美，今年8歲，體重24.3公斤")
```
