```python
import pybryt
```

```python
# 正確解答
def clean_sensor_data(readings):
    return [r for r in readings if r["value"] is not None]
```

```python
result1 = clean_sensor_data([
    {"sensor": "溫度A", "value": 25.3},
    {"sensor": "溫度B", "value": None},
    {"sensor": "濕度A", "value": 60},
])

result2 = clean_sensor_data([
    {"sensor": "X", "value": None},
    {"sensor": "Y", "value": None},
])

result3 = clean_sensor_data([
    {"sensor": "Z1", "value": 0},
    {"sensor": "Z2", "value": 100},
])

pybryt.Value(result1, name="混合空值過濾",
             success_message="正確過濾None，保留其他數值",
             failure_message="應保留 [溫度A:25.3, 濕度A:60]，過濾掉溫度B")

pybryt.Value(result2, name="全部為空值",
             success_message="正確處理全空的情況",
             failure_message="全部為None時應回傳空list []")

pybryt.Value(result3, name="數值0不應被誤判為空值",
             success_message="正確保留數值0",
             failure_message="value=0 是合法數值不應被過濾（常見錯誤：用 if value 判斷而非 if value is not None）")
```
