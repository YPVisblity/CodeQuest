```python
import pybryt
```

```python
# 正確解答
def average_sensor_value(readings):
    valid = [r["value"] for r in readings if r["value"] is not None]
    if len(valid) == 0:
        return "Error:無有效資料"
    return round(sum(valid) / len(valid), 2)
```

```python
result1 = average_sensor_value([
    {"sensor": "溫度A", "value": 20},
    {"sensor": "溫度B", "value": None},
    {"sensor": "溫度C", "value": 30},
])

result2 = average_sensor_value([
    {"sensor": "X", "value": None},
    {"sensor": "Y", "value": None},
])

result3 = average_sensor_value([
    {"sensor": "Z1", "value": 10},
    {"sensor": "Z2", "value": 15},
    {"sensor": "Z3", "value": 20},
])

pybryt.Value(result1, name="混合空值後平均",
             success_message="正確過濾並計算平均",
             failure_message="過濾None後應為 (20+30)/2 = 25.0")

pybryt.Value(result2, name="全空值除以零防護",
             success_message="正確處理除以零的邊界情況",
             failure_message="全部為None時應回傳 'Error:無有效資料'，而非讓程式崩潰")

pybryt.Value(result3, name="正常三筆平均",
             success_message="正確計算平均值",
             failure_message="應為 (10+15+20)/3 = 15.0")
```
