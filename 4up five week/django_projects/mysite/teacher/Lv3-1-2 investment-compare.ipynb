```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pybryt
```

```python
# 正確解答
def plot_compare_investment(months,stock_values,fund_values):
    fig, ax = plt.subplots()
    ax.plot(months,stock_values,label="股票")
    ax.plot(months,fund_values,label="基金")
    ax.legend()
    return ax
```

```python
result_ax = plot_compare_investment(
    [1, 2, 3, 4],
    [100, 120, 90, 150],
    [80, 85, 95, 110],
)
legend1 = result_ax.get_legend()
legend_labels1 = [t.get_text() for t in legend1.get_texts()] if legend1 else []
line_count1 = len(result_ax.get_lines())

result_ax2 = plot_compare_investment(
    [1, 2, 3],
    [50, 60, 55],
    [70, 65, 80],
)
legend2 = result_ax2.get_legend()
legend_labels2 = [t.get_text() for t in legend2.get_texts()] if legend2 else []

pybryt.Value(legend_labels1, name="圖例標籤正確",
             success_message="圖例正確標示股票與基金",
             failure_message="圖例應顯示 ['股票', '基金']，記得幫每條線加 label 並呼叫 ax.legend()")

pybryt.Value(line_count1, name="畫了兩條線",
             success_message="正確畫出股票與基金兩條線",
             failure_message="應該要在同一張圖上畫出股票和基金兩條線")

pybryt.Value(legend_labels2, name="第二組資料圖例仍正確",
             success_message="換一組資料後圖例標籤仍正確",
             failure_message="圖例應顯示 ['股票', '基金']")
```
