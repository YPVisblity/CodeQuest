```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pybryt
```

```python
# 正確解答
def create_pie_chart(activity, hours):
    fig, ax = plt.subplots()

    ax.pie(hours, labels=activity)
    ax.set_title("一天時間分配")

    return ax
```

```python
# 第一組資料
result_ax = create_pie_chart(
    ["睡覺", "上課", "寫作業", "休閒"],
    [8, 6, 4, 6]
)

count1 = len(result_ax.patches)

labels1 = [
    text.get_text()
    for text in result_ax.texts
    if text.get_text()
]

ratios1 = [
    round((wedge.theta2 - wedge.theta1) / 360, 6)
    for wedge in result_ax.patches
]

title_result = result_ax.get_title()

# 第二組資料
result_ax2 = create_pie_chart(
    ["讀書", "運動"],
    [3, 1]
)

count2 = len(result_ax2.patches)

labels2 = [
    text.get_text()
    for text in result_ax2.texts
    if text.get_text()
]

ratios2 = [
    round((wedge.theta2 - wedge.theta1) / 360, 6)
    for wedge in result_ax2.patches
]

pybryt.Value(
    count1,
    name="第一組圓餅區塊數量",
    success_message="正確畫出四個區塊",
    failure_message="第一組資料應畫出四個區塊"
)

pybryt.Value(
    labels1,
    name="第一組活動標籤",
    success_message="活動標籤正確",
    failure_message="請使用 activity 作為圓餅圖標籤"
)

pybryt.Value(
    ratios1,
    name="第一組時間比例",
    success_message="各活動所占比例正確",
    failure_message="請使用 hours 決定圓餅圖比例"
)

pybryt.Value(
    count2,
    name="第二組圓餅區塊數量",
    success_message="更換資料後區塊數量仍正確",
    failure_message="請勿將區塊數量寫死"
)

pybryt.Value(
    labels2,
    name="第二組活動標籤",
    success_message="第二組標籤正確",
    failure_message="請勿將活動標籤寫死"
)

pybryt.Value(
    ratios2,
    name="第二組時間比例",
    success_message="第二組時間比例正確",
    failure_message="請勿將時間比例寫死"
)

pybryt.Value(title_result, name="圖表標題")
```
