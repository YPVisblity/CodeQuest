```python
import pybryt
```

```python
from bs4 import BeautifulSoup

def extract_titles(html):
    soup = BeautifulSoup(html, "html.parser")
    h2_tags = soup.find_all("h2")
    return [tag.get_text() for tag in h2_tags]
```

```python
html1 = """
<html><body>
<h2>第一篇文章</h2>
<p>內容...</p>
<h2>第二篇文章</h2>
<p>內容...</p>
<h2>第三篇文章</h2>
</body></html>
"""
result1 = extract_titles(html1)

html2 = """
<html><body>
<h1>這不是h2</h1>
<h2>唯一的標題</h2>
</body></html>
"""
result2 = extract_titles(html2)

html3 = "<html><body><p>沒有任何h2標籤</p></body></html>"
result3 = extract_titles(html3)

pybryt.Value(result1, name="多個h2標籤",
             success_message="正確抓取多個h2並保持順序",
             failure_message="應回傳 ['第一篇文章', '第二篇文章', '第三篇文章']")

pybryt.Value(result2, name="混合標籤過濾",
             success_message="正確過濾非h2標籤",
             failure_message="應只回傳 ['唯一的標題']，不應包含h1內容")

pybryt.Value(result3, name="無h2標籤情況",
             success_message="正確處理空結果",
             failure_message="沒有h2標籤時應回傳空list []")
```
