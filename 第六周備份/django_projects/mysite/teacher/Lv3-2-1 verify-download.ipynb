```python
import pybryt
```

```python
# 正確解答
def verify_download(file_info):
    if not file_info["filename"].endswith(".mp4"):
        return "Error:格式錯誤"
    if file_info["size"] <= 0:
        return "Error:檔案損毀"
    if not file_info["checksum_valid"]:
        return "Error:校驗失敗"
    return "OK:下載成功"
```

```python
result1 = verify_download({"filename": "lesson01.mp4", "size": 15420, "checksum_valid": True})
result2 = verify_download({"filename": "lesson02.txt", "size": 8300, "checksum_valid": True})
result3 = verify_download({"filename": "lesson03.mp4", "size": 0, "checksum_valid": True})
result4 = verify_download({"filename": "lesson04.mp4", "size": 12000, "checksum_valid": False})
result5 = verify_download({"filename": "lesson05.txt", "size": 0, "checksum_valid": False})

pybryt.Value(result1, name="正常下載成功",
             success_message="正確判定完整檔案",
             failure_message="格式、大小、校驗都正確時應回傳 'OK:下載成功'")

pybryt.Value(result2, name="格式錯誤偵測",
             success_message="正確偵測非mp4格式",
             failure_message="非.mp4結尾應回傳 'Error:格式錯誤'")

pybryt.Value(result3, name="檔案損毀偵測",
             success_message="正確偵測檔案大小異常",
             failure_message="size<=0應回傳 'Error:檔案損毀'")

pybryt.Value(result4, name="校驗失敗偵測",
             success_message="正確偵測校驗碼異常",
             failure_message="checksum_valid為False應回傳 'Error:校驗失敗'")

pybryt.Value(result5, name="多重錯誤優先順序",
             success_message="正確依優先順序（格式錯誤>檔案損毀>校驗失敗）回傳",
             failure_message="格式、大小、校驗同時錯誤時，應優先回傳 'Error:格式錯誤'")
```
