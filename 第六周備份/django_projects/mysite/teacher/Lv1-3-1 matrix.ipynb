```python
import pybryt

def matrix_add(A, B):

    rows = len(A)
    cols = len(A[0])

    result = []

    for i in range(rows):

        row = []

        for j in range(cols):
            row.append(A[i][j] + B[i][j])

        result.append(row)

    

    return result
```

```python
A = [
    [1,2],
    [3,4]
]

B = [
    [5,6],
    [7,8]
]

result = matrix_add(A,B)
pybryt.Value(
        result,
        success_message="矩陣相加成功",
        failure_message="矩陣相加失敗"
    )
```

```python
A = [
    [2,1],
    [0,3]
]

B = [
    [1,4],
    [5,2]
]

result = matrix_add(A,B)

pybryt.Value(
    result,
    success_message="不同數值測試成功",
    failure_message="不同數值測試失敗"
)
```
