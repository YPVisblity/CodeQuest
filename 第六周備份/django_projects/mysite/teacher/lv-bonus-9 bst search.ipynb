```python
import pybryt
```

```python
class BSTNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# 範例：插入操作（已完成，供參考）
def insert(root, data):
    if root is None:
        return BSTNode(data)
    if data < root.data:
        root.left = insert(root.left, data)
    elif data > root.data:
        root.right = insert(root.right, data)
    return root

def inorder_values(node, result=None):
    if result is None:
        result = []
    if node is not None:
        inorder_values(node.left, result)
        result.append(node.data)
        inorder_values(node.right, result)
    return result

# 手動建立一棵 BST 來測試
bst_root = None
for value in [8, 3, 10, 1, 6, 14]:
    bst_root = insert(bst_root, value)

print(inorder_values(bst_root))  # 預期輸出：[1, 3, 6, 8, 10, 14]（中序走訪結果應為由小到大排序）
```

```python
def search(root, target):
    """在 BST 中搜尋 target 是否存在，回傳 True 或 False"""
    if root is None:
        return False
    if target == root.data:
        return True
    elif target < root.data:
        return search(root.left, target)
    else:
        return search(root.right, target)


# ===== 自我檢查 =====
assert search(bst_root, 6) == True
assert search(bst_root, 14) == True
assert search(bst_root, 100) == False
assert search(bst_root, 0) == False

print("恭喜！基礎任務全部通過，可以繼續挑戰進階任務。")
```

```python
r1 = search(bst_root, 6)
r2 = search(bst_root, 14)
r3 = search(bst_root, 100)
r4 = search(bst_root, 0)

pybryt.ReturnValue(r1, name="search_found_middle",
    success_message="找到正確的節點6",
    failure_message="search(bst_root,6) 應回傳 True")
pybryt.ReturnValue(r2, name="search_found_leaf",
    success_message="找到正確的葉節點14",
    failure_message="search(bst_root,14) 應回傳 True")
pybryt.ReturnValue(r3, name="search_not_found_large",
    success_message="能正確判斷不存在的較大值100",
    failure_message="search(bst_root,100) 應回傳 False")
pybryt.ReturnValue(r4, name="search_not_found_small",
    success_message="能正確判斷不存在的較小值0",
    failure_message="search(bst_root,0) 應回傳 False")
```
