```python
import pybryt
```

```python
def matrix_sum(matrix):
    total = 0
    for row in matrix:
        for number in row:
            total += number
    return total

pybryt.ReturnValue(matrix_sum([[1, 2], [3, 4]]), success_message="matrix sum 2x2 passed", failure_message="matrix sum [[1,2],[3,4]] should return 10")
pybryt.ReturnValue(matrix_sum([[5, 0, 2], [1, 3, 4]]), success_message="matrix sum 2x3 passed", failure_message="matrix sum [[5,0,2],[1,3,4]] should return 15")
pybryt.ReturnValue(matrix_sum([[10], [20], [30]]), success_message="matrix sum one column passed", failure_message="matrix sum [[10],[20],[30]] should return 60")
```
