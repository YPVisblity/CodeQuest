```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pybryt
```

```python
# 正確解答
def create_bar_chart(fruit, price):
    fig, ax = plt.subplots()

    ax.bar(fruit, price)
    ax.set_title("水果價格")
    ax.set_xlabel("水果")
    ax.set_ylabel("價錢")

    return ax
```

```python
# 第一組資料
result_ax = create_bar_chart(
    ["Apple", "Banana", "Orange", "Grape"],
    [30, 45, 25, 40]
)

count1 = len(result_ax.patches)
price_result1 = [
    bar.get_height()
    for bar in result_ax.patches
]
fruit_result1 = [
    label.get_text()
    for label in result_ax.get_xticklabels()
]

title_result = result_ax.get_title()
xlabel_result = result_ax.get_xlabel()
ylabel_result = result_ax.get_ylabel()

# 第二組資料
result_ax2 = create_bar_chart(
    ["Mango", "Peach"],
    [50, 35]
)

count2 = len(result_ax2.patches)
price_result2 = [
    bar.get_height()
    for bar in result_ax2.patches
]
fruit_result2 = [
    label.get_text()
    for label in result_ax2.get_xticklabels()
]

pybryt.Value(
    count1,
    name="第一組長條數量",
    success_message="正確畫出四個長條",
    failure_message="第一組資料應畫出四個長條"
)

pybryt.Value(
    price_result1,
    name="第一組長條高度",
    success_message="長條高度正確",
    failure_message="請使用傳入的 price 作為長條高度"
)

pybryt.Value(
    fruit_result1,
    name="第一組水果名稱",
    success_message="水果名稱正確",
    failure_message="請使用傳入的 fruit 作為分類"
)

pybryt.Value(
    count2,
    name="第二組長條數量",
    success_message="更換資料後仍能正確繪圖",
    failure_message="請勿將長條數量寫死"
)

pybryt.Value(
    price_result2,
    name="第二組長條高度",
    success_message="第二組長條高度正確",
    failure_message="請勿將價格寫死"
)

pybryt.Value(
    fruit_result2,
    name="第二組水果名稱",
    success_message="第二組水果名稱正確",
    failure_message="請勿將水果名稱寫死"
)

pybryt.Value(title_result, name="圖表標題")
pybryt.Value(xlabel_result, name="X軸名稱")
pybryt.Value(ylabel_result, name="Y軸名稱")
```
