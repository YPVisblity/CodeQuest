```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pybryt
```

```python
# 正確解答
def create_plot(days, temperatures):
    fig, ax = plt.subplots()

    ax.plot(days, temperatures)
    ax.set_title("溫差")
    ax.set_xlabel("天數")
    ax.set_ylabel("溫度")

    return ax
```

```python
#第一組的資料
result_ax = create_plot(
    [1, 2, 3, 4, 5],
    [22, 24, 23, 26, 28]
)

line_count = len(result_ax.lines)
line = result_ax.lines[0]

x_data = list(line.get_xdata())
y_data = list(line.get_ydata())
plot_title = result_ax.get_title()
x_label = result_ax.get_xlabel()
y_label = result_ax.get_ylabel()

# 第二組資料
result_ax2 = create_plot(
    [1, 2, 3],
    [18, 21, 20]
)

line2 = result_ax2.lines[0]
x_data2 = list(line2.get_xdata())
y_data2 = list(line2.get_ydata())

pybryt.Value(
    line_count,
    name="折線數量",
    success_message="已正確畫出一條折線",
    failure_message="圖表中應該只有一條折線"
)

pybryt.Value(
    x_data,
    name="第一組X軸資料",
    success_message="第一組 X 軸資料正確",
    failure_message="折線的 X 軸資料不正確"
)

pybryt.Value(
    y_data,
    name="第一組Y軸資料",
    success_message="第一組 Y 軸資料正確",
    failure_message="折線的 Y 軸資料不正確"
)

pybryt.Value(
    x_data2,
    name="第二組X軸資料未寫死",
    success_message="更換資料後仍能正確繪圖",
    failure_message="請使用函式收到的 days，不要把資料寫死"
)

pybryt.Value(
    y_data2,
    name="第二組Y軸資料未寫死",
    success_message="更換溫度資料後仍能正確繪圖",
    failure_message="請使用函式收到的 temperatures，不要把資料寫死"
)

pybryt.Value(plot_title, name="圖表標題")
pybryt.Value(x_label, name="X軸名稱")
pybryt.Value(y_label, name="Y軸名稱")
```
