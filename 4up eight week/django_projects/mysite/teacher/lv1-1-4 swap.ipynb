```python
import pybryt
```

```python
def swap_values(a,b):
    a,b = b,a
    return (a,b)
```

```python
result1 = swap_values(1,2)
result2 = swap_values(100,-50)
result3 = swap_values("蘋果","香蕉")

pybryt.Value(result1, name="交換整數(1,2)",
             success_message="數值交換正確",
             failure_message="swap_values(1,2) 應回傳 (2,1)")

pybryt.Value(result2, name="交換負數(100,-50)",
             success_message="正負數交換正確",
             failure_message="swap_values(100,-50) 應回傳 (-50,100)")

pybryt.Value(result3, name="交換字串",
             success_message="字串交換正確",
             failure_message="swap_values('蘋果','香蕉') 應回傳 ('香蕉','蘋果')")
```
