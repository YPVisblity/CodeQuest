```python
import pybryt
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return None

    def is_empty(self):
        return len(self.items) == 0


class QueueUsingStacks:
    def __init__(self):
        self.stack_in = Stack()
        self.stack_out = Stack()

    def enqueue(self, item):
        self.stack_in.push(item)

    def dequeue(self):
        # 只有在stack_out空了的時候，才把 stack_in 整批倒過去
        # 確保先進的元素被倒到stack_out之後會先被 pop
        if self.stack_out.is_empty():
            while not self.stack_in.is_empty():
                self.stack_out.push(self.stack_in.pop())
        return self.stack_out.pop()
```

```python
qs = QueueUsingStacks()

# 結構檢查:必須是用提供的Stack，不能被繞過
assert isinstance(qs.stack_in, Stack), "stack_in必須是Stack的實例"
assert isinstance(qs.stack_out, Stack), "stack_out必須是Stack的實例"

qs.enqueue(1)
qs.enqueue(2)
qs.enqueue(3)

# 結構檢查:enqueue之後元素應該都還在stack_in，
# stack_out應該是空的，確保不是每次都硬轉移
assert qs.stack_out.is_empty(),"尚未dequeue前stack_out應為空"

r1 = qs.dequeue()
r2_precheck = qs.stack_in.is_empty()  # 轉移後 stack_in 應該被清空
qs.enqueue(4)
r2 = qs.dequeue()
r3 = qs.dequeue()
r4 = qs.dequeue()
```

```python
pybryt.ReturnValue(r1, name="dequeue_1",
    success_message="第一次dequeue正確(FIFO 先進先出)",
    failure_message="dequeue() 順序錯誤，應該先出的是最早 enqueue 的元素")

pybryt.ReturnValue(r2, name="dequeue_2",
    success_message="延遲轉移邏輯正確",
    failure_message="stack_out空了之後，沒有正確把 stack_in 的元素轉移過去")

pybryt.ReturnValue(r3, name="dequeue_3",
    success_message="延遲轉移邏輯正確",
    failure_message="stack_out空了之後，沒有正確把 stack_in 的元素轉移過去")

pybryt.ReturnValue(r4, name="dequeue_4",
    success_message="延遲轉移邏輯正確",
    failure_message="stack_out空了之後，沒有正確把 stack_in 的元素轉移過去")

pybryt.Value(True, name="uses_two_stacks",
    success_message="確實使用了兩個Stack實作",
    failure_message="未偵測到使用提供的Stack類別，或延遲轉移邏輯不正確")
```
